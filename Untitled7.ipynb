```python
student_scores = {"Alice":95,"bob": 82,"charlie": 78,"david": 60,"eve": 55}
for student,score in student_scores.items():
  if score >= 90:
    grade = "A"
  elif score >= 80:
    grade = "B"
  elif score >= 70:
    grade = "C"
  elif score >= 60:
    grade = "D"
  else:
    grade = "F"
  print(f"{student} received grade {grade}")
```
