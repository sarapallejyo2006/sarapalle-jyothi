```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Input
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# Linux path pointing to your uploaded and extracted dataset
dataset_path = "/content/dataset"

# Image Preprocessing
datagen = ImageDataGenerator(
    rescale=1./255,
    validation_split=0.2
)

# Training Data
train_generator = datagen.flow_from_directory(
    dataset_path,
    target_size=(128,128),
    batch_size=32,
    class_mode='binary',
    subset='training'
)

# Validation Data
validation_generator = datagen.flow_from_directory(
    dataset_path,
    target_size=(128,128),
    batch_size=32,
    class_mode='binary',
    subset='validation'
)

# CNN Model Architecture
model = Sequential()

# Add an Input layer explicitly as recommended by Keras
model.add(Input(shape=(128, 128, 3)))

# Convolution Layer 1
model.add(Conv2D(32, (3,3), activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2)))

# Convolution Layer 2
model.add(Conv2D(64, (3,3), activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2)))

# Flatten Layer
model.add(Flatten())

# Dense Layer
model.add(Dense(128, activation='relu'))

# Output Layer
model.add(Dense(1, activation='sigmoid'))

# Compile Model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train Model
history = model.fit(
    train_generator,
    validation_data=validation_generator,
    epochs=10
)

# Save the trained model file directly to the cloud workspace
model.save("face_mask_detector.keras")
print("Model Trained and Saved Successfully!")
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# READ IMAGE
# =====================================================

# IMPORTANT: The original path 'C:/Users/Edunet Foundation/...' is a local Windows path
# and does not exist in the Colab environment. You need to upload your image
# to Colab or use an existing one. For demonstration, we'll use an image
# from the previously uploaded 'data.zip' dataset.
image = cv2.imread(
    "/content/WhatsApp Image 2026-08-22 at 4.17.31 PM.jpeg" # Example path, replace with your image path
)

# Check if the image was loaded successfully
if image is None:
    print("Error: Image not loaded. Please check the image path.")
else:
    # =====================================================
    # GRAYSCALE
    # =====================================================

    gray = cv2.cvtColor(
        image,
        cv2.COLOR_BGR2GRAY
    )

    # =====================================================
    # GAUSSIAN NOISE
    # =====================================================

    gaussian = gray + np.random.normal(
        0,
        25,
        gray.shape
    )

    gaussian = np.clip(
        gaussian,
        0,
        255
    ).astype(np.uint8)

    # =====================================================
    # SALT & PEPPER NOISE
    # =====================================================

    sp = gray.copy()

    prob = 0.02

    salt = np.random.rand(*gray.shape) < prob
    sp[salt] = 255

    pepper = np.random.rand(*gray.shape) < prob
    sp[pepper] = 0

    # =====================================================
    # GAUSSIAN BLUR
    # =====================================================

    blur = cv2.GaussianBlur(
        gaussian,
        (5,5),
        0
    )

    # =====================================================
    # SOBEL X
    # =====================================================

    gx = cv2.Sobel(
        blur,
        cv2.CV_64F,
        1,
        0,
        ksize=3
    )

    # =====================================================
    # SOBEL Y
    # =====================================================

    gy = cv2.Sobel(
        blur,
        cv2.CV_64F,
        0,
        1,
        ksize=3
    )

    # =====================================================
    # GRADIENT MAGNITUDE
    # =====================================================

    gradient = np.sqrt(
        gx**2 + gy**2
    )

    # =====================================================
    # CANNY
    # =====================================================

    edges = cv2.Canny(
        blur,
        100,
        200
    )

    # =====================================================
    # DISPLAY
    # =====================================================

    plt.figure(figsize=(24,12))

    # 1 Original
    plt.subplot(3,3,1)
    plt.imshow(cv2.cvtColor(image,cv2.COLOR_BGR2RGB))
    plt.title("Original Image")
    plt.axis("off")

    # 2 Gray
    plt.subplot(3,3,2)
    plt.imshow(gray,cmap="gray")
    plt.title("Grayscale")
    plt.axis("off")

    # 3 Gaussian Noise
    plt.subplot(3,3,3)
    plt.imshow(gaussian,cmap="gray")
    plt.title("Gaussian Noise")
    plt.axis("off")

    # 4 Salt & Pepper
    plt.subplot(3,3,4)
    plt.imshow(sp,cmap="gray")
    plt.title("Salt & Pepper Noise")
    plt.axis("off")

    # 5 Blur
    plt.subplot(3,3,5)
    plt.imshow(blur,cmap="gray")
    plt.title("Gaussian Blur")
    plt.axis("off")

    # 6 Sobel X
    plt.subplot(3,3,6)
    plt.imshow(np.abs(gx),cmap="gray")
    plt.title("Sobel Gx")
    plt.axis("off")

    # 7 Sobel Y
    plt.subplot(3,3,7)
    plt.imshow(np.abs(gy),cmap="gray")
    plt.title("Sobel Gy")
    plt.axis("off")

    # 8 Gradient
    plt.subplot(3,3,8)
    plt.imshow(gradient,cmap="gray")
    plt.title("Gradient Magnitude")
    plt.axis("off")

    # 9 Canny
    plt.subplot(3,3,9)
    plt.imshow(edges,cmap="gray")
    plt.title("Canny Edge Detection")
    plt.axis("off")

    plt.tight_layout()
    plt.show()
```
